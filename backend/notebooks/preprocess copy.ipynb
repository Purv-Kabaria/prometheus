```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# 1. Load Data
df = pd.read_csv('hackathon_final_data.csv')

# 2. Parse Datetime
# We use 'Start time UTC' and convert it to standard datetime
df['datetime'] = pd.to_datetime(df['Start time UTC'])
df = df.sort_values('datetime').set_index('datetime')

# Rename for easier typing
df = df.rename(columns={'Electricity consumption (MWh)': 'consumption'})

# Drop the old string columns
df = df.drop(columns=['Start time UTC', 'End time UTC'])

# 3. Extract Temporal Features
df['hour'] = df.index.hour
df['dayofweek'] = df.index.dayofweek
df['month'] = df.index.month
df['is_weekend'] = df['dayofweek'].isin([5, 6]).astype(int)
df['date'] = df.index.date

print("Data loaded! Shape:", df.shape)
print("Date Range:", df.index.min(), "to", df.index.max())
print("\nMissing Values:\n", df.isnull().sum())
```
